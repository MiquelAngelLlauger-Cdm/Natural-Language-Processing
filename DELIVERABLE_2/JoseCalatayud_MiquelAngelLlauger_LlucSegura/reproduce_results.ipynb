# Reproduce results

Loads the supplied data and fitted artifacts without training, then reports train/test non-O accuracy, weighted F1, confusion matrices, and TINY TEST predictions in the required word/tag format.

If an optional artifact is absent or incomplete, the notebook identifies it explicitly and continues with the available models.


```python
# === Setup (works locally and on Google Colab) ===
import sys, os
IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB:
    !pip install -q sklearn-crfsuite transformers seqeval
    DATA_DIR = 'data'
else:
    DATA_DIR = 'data'
FITTED_DIR = 'fitted_models'

import warnings; warnings.filterwarnings('ignore')
import re, pickle, numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.metrics import f1_score, accuracy_score, confusion_matrix
sns.set_theme(style='whitegrid')
```

## 1. Load data

```python
def load_data(path):
    df = pd.read_csv(path)
    df['words'] = df['words'].astype(str)
    df['sentence_id'] = df['sentence_id'].astype(int)
    return [list(zip(g['words'].tolist(), g['tags'].tolist()))
            for _, g in df.groupby('sentence_id', sort=True)]

train_sents = load_data(os.path.join(DATA_DIR, 'train_data_ner.csv'))
test_sents  = load_data(os.path.join(DATA_DIR, 'test_data_ner.csv'))
tiny_sents  = load_data(os.path.join(DATA_DIR, 'tiny_test.csv'))

y_train = [[t for _, t in s] for s in train_sents]
y_test  = [[t for _, t in s] for s in test_sents]
y_tiny  = [[t for _, t in s] for s in tiny_sents]
print(f'train: {len(train_sents)} | test: {len(test_sents)} | tiny: {len(tiny_sents)}')
```

## 2. Metric helpers (shared by all models)

```python
def flatten(seq): return [x for sub in seq for x in sub]

def non_o_accuracy(y_true, y_pred):
    """Accuracy computed only where the GROUND TRUTH tag is not 'O'."""
    yt, yp = flatten(y_true), flatten(y_pred)
    idx = [k for k, t in enumerate(yt) if t != 'O']
    return accuracy_score([yt[k] for k in idx], [yp[k] for k in idx])

def weighted_f1(y_true, y_pred):
    """Weighted F1 over the non-'O' labels present in the ground truth."""
    yt, yp = flatten(y_true), flatten(y_pred)
    labels = sorted(set(yt) - {'O'})
    return f1_score(yt, yp, average='weighted', labels=labels, zero_division=0)

def plot_cm(y_true, y_pred, title, cmap='Blues'):
    yt, yp = flatten(y_true), flatten(y_pred)
    labels = sorted(set(yt) - {'O'})
    cm = confusion_matrix(yt, yp, labels=labels)
    plt.figure(figsize=(9,7))
    sns.heatmap(cm, annot=True, fmt='d', cmap=cmap, xticklabels=labels, yticklabels=labels)
    plt.title(title); plt.ylabel('True'); plt.xlabel('Predicted'); plt.show()

def print_tiny(sents, y_pred, name):
    print(f'=== TINY TEST â€” {name} ===')
    for s, pred in zip(sents, y_pred):
        print(' '.join(f'{w}/{t}' for (w, _), t in zip(s, pred)))
    print(f'Tiny-test accuracy (non-O): {non_o_accuracy(y_tiny, y_pred):.4f}\n')
```

## 3. Load models from `fitted_models/`

Each model needs the same pre-processing it was trained with, so we re-declare the CRF
feature templates and the BiLSTM architecture here (kept identical to the training
notebooks).

```python
# ---- CRF feature templates (identical to 02_Model_CRF.ipynb) ----
def word_shape(w):
    s = re.sub('[A-Z]', 'X', w); s = re.sub('[a-z]', 'x', s); s = re.sub('[0-9]', 'd', s)
    return s

def feat_v1(sent, i):
    w = str(sent[i][0])
    return {'bias': 1.0, 'w.lower': w.lower(), 'w.isupper': w.isupper(),
            'w.istitle': w.istitle(), 'w.isdigit': w.isdigit()}

def feat_v2(sent, i):
    w = str(sent[i][0])
    f = {'bias': 1.0, 'w.lower': w.lower(), 'w[-3:]': w[-3:], 'w[-2:]': w[-2:], 'w[:3]': w[:3],
         'w.isupper': w.isupper(), 'w.istitle': w.istitle(), 'w.isdigit': w.isdigit(),
         'w.shape': word_shape(w)}
    if i > 0:
        p = str(sent[i-1][0]); f.update({'-1:lower': p.lower(), '-1:istitle': p.istitle(), '-1:isupper': p.isupper()})
    else: f['BOS'] = True
    if i < len(sent)-1:
        n = str(sent[i+1][0]); f.update({'+1:lower': n.lower(), '+1:istitle': n.istitle(), '+1:isupper': n.isupper()})
    else: f['EOS'] = True
    return f

def feat_v3(sent, i):
    w = str(sent[i][0])
    f = {'bias': 1.0, 'w.lower': w.lower(), 'w[-4:]': w[-4:], 'w[-3:]': w[-3:], 'w[-2:]': w[-2:],
         'w[:2]': w[:2], 'w[:3]': w[:3], 'w.isupper': w.isupper(), 'w.istitle': w.istitle(),
         'w.isdigit': w.isdigit(), 'w.shape': word_shape(w), 'w.len': len(w),
         'w.has_hyphen': '-' in w, 'w.has_digit': any(ch.isdigit() for ch in w),
         'w.is_punct': not any(ch.isalnum() for ch in w)}
    for off in (-2, -1, 1, 2):
        j = i + off
        if 0 <= j < len(sent):
            c = str(sent[j][0])
            f.update({f'{off:+d}:lower': c.lower(), f'{off:+d}:istitle': c.istitle(),
                      f'{off:+d}:isupper': c.isupper(), f'{off:+d}:shape': word_shape(c)})
    if i == 0: f['BOS'] = True
    if i == len(sent)-1: f['EOS'] = True
    return f

CRF_FEATURES = {'CRF v1': feat_v1, 'CRF v2': feat_v2, 'CRF v3': feat_v3}
CRF_FILES    = {'CRF v1': 'crf_v1.pkl', 'CRF v2': 'crf_v2.pkl', 'CRF v3': 'crf_v3.pkl'}

crfs = {}
for name, fn in CRF_FILES.items():
    path = os.path.join(FITTED_DIR, fn)
    if os.path.exists(path):
        with open(path, 'rb') as fh: crfs[name] = pickle.load(fh)
print('Loaded CRFs:', list(crfs))

def crf_predict(crf, sents, featfn):
    X = [[featfn(s, i) for i in range(len(s))] for s in sents]
    return crf.predict(X)
```

```python
# ---- BiLSTM (identical architecture to 03_Model_DL_BiLSTM.ipynb) ----
import torch, torch.nn as nn
device = torch.device('cuda' if torch.cuda.is_available()
                      else 'mps' if torch.backends.mps.is_available() else 'cpu')

class BiLSTMTagger(nn.Module):
    def __init__(self, vocab_size, tagset_size, embedding_dim, hidden_dim, dropout):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)
        self.lstm = nn.LSTM(embedding_dim, hidden_dim, batch_first=True, bidirectional=True)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_dim*2, tagset_size)
    def forward(self, x):
        e = self.dropout(self.embedding(x))
        h, _ = self.lstm(e)
        return self.fc(self.dropout(h))

bilstm = None
map_path = os.path.join(FITTED_DIR, 'dl_mappings.pkl')
w_path   = os.path.join(FITTED_DIR, 'bilstm_model.pth')
if os.path.exists(map_path) and os.path.exists(w_path):
    with open(map_path, 'rb') as fh: M = pickle.load(fh)
    word_to_ix, ix_to_tag, HP = M['word_to_ix'], M['ix_to_tag'], M['hp']
    bilstm = BiLSTMTagger(len(word_to_ix), len(M['tag_to_ix']),
                          HP['embedding_dim'], HP['hidden_dim'], HP['dropout']).to(device)
    bilstm.load_state_dict(torch.load(w_path, map_location=device))
    bilstm.eval()
    print('Loaded BiLSTM.')

    @torch.no_grad()
    def bilstm_predict(sents):
        preds = []
        for sent in sents:
            ids = [word_to_ix.get(w, word_to_ix['<UNK>']) for w, _ in sent]
            x = torch.tensor(ids).unsqueeze(0).to(device)
            p = bilstm(x).argmax(-1).squeeze(0).tolist()
            preds.append([ix_to_tag.get(i, 'O') for i in p])
        return preds
else:
    print('BiLSTM artifact not found â€” skipping.')
```

```python
# ---- DistilBERT (optional, loaded only if the artifact exists) ----
bert = None
bert_dir = os.path.join(FITTED_DIR, 'distilbert_ner')
lbl_path = os.path.join(FITTED_DIR, 'distilbert_labels.pkl')
if os.path.isdir(bert_dir) and os.path.exists(lbl_path):
    from transformers import AutoTokenizer, AutoModelForTokenClassification
    with open(lbl_path, 'rb') as fh: L = pickle.load(fh)
    bert_id2label, BERT_MAX = L['id2label'], L['max_len']
    bert_tok = AutoTokenizer.from_pretrained(bert_dir)
    bert = AutoModelForTokenClassification.from_pretrained(bert_dir).to(device).eval()
    print('Loaded DistilBERT.')

    @torch.no_grad()
    def bert_predict(sents, batch_size=32):
        preds = []
        for start in range(0, len(sents), batch_size):
            chunk = sents[start:start+batch_size]
            batch_words = [[w for w, _ in s] for s in chunk]
            enc = bert_tok(batch_words, is_split_into_words=True, truncation=True,
                           max_length=BERT_MAX, padding=True, return_tensors='pt').to(device)
            pred_ids = bert(**enc).logits.argmax(-1).cpu().tolist()
            for b, s in enumerate(chunk):
                word_ids = enc.word_ids(b); seq, prev = [], None
                for pos, wid in enumerate(word_ids):
                    if wid is not None and wid != prev:
                        seq.append(bert_id2label[pred_ids[b][pos]])
                    prev = wid
                words = [w for w, _ in s]
                if len(seq) < len(words): seq += ['O'] * (len(words) - len(seq))
                preds.append(seq[:len(words)])
        return preds
else:
    print('DistilBERT artifact not found â€” skipping (re-run 04 on Colab GPU to add it).')
```

## 4. Compute predictions for every available model

```python
# Collect a (name -> {train,test,tiny} predictions) dict
PRED = {}
for name, featfn in CRF_FEATURES.items():
    if name in crfs:
        PRED[name] = {
            'train': crf_predict(crfs[name], train_sents, featfn),
            'test':  crf_predict(crfs[name], test_sents,  featfn),
            'tiny':  crf_predict(crfs[name], tiny_sents,  featfn),
        }
if bilstm is not None:
    PRED['BiLSTM'] = {'train': bilstm_predict(train_sents),
                      'test':  bilstm_predict(test_sents),
                      'tiny':  bilstm_predict(tiny_sents)}
if bert is not None:
    PRED['DistilBERT'] = {'train': bert_predict(train_sents),
                          'test':  bert_predict(test_sents),
                          'tiny':  bert_predict(tiny_sents)}
print('Models evaluated:', list(PRED))
```

## 5. Train / Test metrics â€” non-`O` accuracy and weighted F1

```python
rows = []
for name, P in PRED.items():
    rows.append({'model': name,
                 'train_acc_nonO': non_o_accuracy(y_train, P['train']),
                 'test_acc_nonO':  non_o_accuracy(y_test,  P['test']),
                 'train_f1':       weighted_f1(y_train, P['train']),
                 'test_f1':        weighted_f1(y_test,  P['test'])})
results = pd.DataFrame(rows).set_index('model').round(4)
display(results)

best = results['test_f1'].idxmax()
print('Best model by test F1:', best)
```

## 6. Confusion matrices â€” best model (train and test)

```python
plot_cm(y_train, PRED[best]['train'], f'{best} â€” TRAIN confusion matrix (excluding O)', cmap='Greens')
plot_cm(y_test,  PRED[best]['test'],  f'{best} â€” TEST confusion matrix (excluding O)',  cmap='Blues')
```

## 7. TINY TEST predictions (`w/t` format) for every model

```python
for name, P in PRED.items():
    print_tiny(tiny_sents, P['tiny'], name)
```
